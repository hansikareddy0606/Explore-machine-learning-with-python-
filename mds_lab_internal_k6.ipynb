```python
import pandas as pd
df = pd.DataFrame({
    'name': ['A', 'B', 'C', 'D', 'E', 'F', 'G'],
    'marks': [45, 50, 55, 60, 65, 70, 200]
})
Q1 = df['marks'].quantile(0.25)
Q3 = df['marks'].quantile(0.75)
IQR = Q3 - Q1
lower_limit = Q1 - 1.5 * IQR
upper_limit = Q3 + 1.5 * IQR
outliers = df[
    (df['marks'] < lower_limit) |
    (df['marks'] > upper_limit)
]

print("Outliers:")
print(outliers)
```

```python

import pandas as pd
df = pd.DataFrame({
    'Maths': [80, 90, 70],
})
print("Median")
print(df.median())
print("mean")
print(df.mean())
print("mode")
print(df.mode())
print(df.max())
print(df.min())
print(df.var())
print("\nStudents with Maths > 80:")
print(df[df['Maths'] > 80])
```

```python
import pandas as pd
df = pd.DataFrame({
    'Name': ['Asha', 'Ravi', 'maya', 'Priya', 'John'],
    'Marks': [85, None, 90, None, 75]
})

print("Original DataFrame:")
print(df)

print("\nMissing Values:")
print(df.isnull())
print("\nAfter Dropping Missing Values:")
print(df.dropna())
print(df.fillna(0))

```

```python
import pandas as pd
df = pd.DataFrame({
    'Name': ['Ravi', 'Priya', 'Amit', 'Sara'],
    'Marks': [80, 65, 90, 45]
})
df['Marks'] = df['Marks'].map(lambda x: x + 5)

df['Name'] = df['Name'].apply(lambda x: x.upper())

print("\nTransformed DataFrame:")
print(df)
```
