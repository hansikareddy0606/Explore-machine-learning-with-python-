```python
import time
import numpy as np

# Define the size of the arrays/lists
size = 10**6

# --- Python List Comparison ---

# Create two large Python lists
python_list1 = list(range(size))
python_list2 = list(range(size))

start_time = time.time()
python_result_list = [a + b for a, b in zip(python_list1, python_list2)]
end_time = time.time()
python_list_time = end_time - start_time

print(f"Time taken for Python list addition: {python_list_time:.6f} seconds")

# --- NumPy Array Comparison ---

# Create two large NumPy arrays
numpy_array1 = np.arange(size)
numpy_array2 = np.arange(size)

start_time = time.time()
numpy_result_array = numpy_array1 + numpy_array2
end_time = time.time()
numpy_array_time = end_time - start_time

print(f"Time taken for NumPy array addition: {numpy_array_time:.6f} seconds")

print(f"\nNumPy is approximately {python_list_time / numpy_array_time:.2f} times faster than Python lists for this operation.")
```

```python
import sys

# --- Python List Memory Usage ---

# Calculate memory for a single integer in a Python list
# (Each integer object in a Python list has overhead)
int_size_python = sys.getsizeof(python_list1[0])

# Approximate total memory for the Python list of integers
# This is a rough estimate as it doesn't account for list structure overhead itself,
# but focuses on the elements.
python_list_memory = sys.getsizeof(python_list1) + size * int_size_python
print(f"Approximate memory used by Python list (size {size}): {python_list_memory / (1024**2):.2f} MB")

# --- NumPy Array Memory Usage ---

# NumPy arrays store elements directly and efficiently
numpy_array_memory = numpy_array1.nbytes
print(f"Memory used by NumPy array (size {size}): {numpy_array_memory / (1024**2):.2f} MB")

print(f"\nFor {size} elements, NumPy uses approximately {python_list_memory / numpy_array_memory:.2f} times less memory than Python lists.")
```
